```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Dataset Creation
students = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']
marks = [78, 85, 92, 67, 88, 95, 72, 81, 90, 76]

df = pd.DataFrame({'Student': students, 'Marks': marks})

# 2. Find the highest scorer
highest_idx = df['Marks'].idxmax()
highest_student = df.loc[highest_idx, 'Student']
highest_score = df.loc[highest_idx, 'Marks']

print(f"Highest Scorer: Student {highest_student} with {highest_score} marks")

# 3. Plot 1: Initial graphical display
plt.figure(figsize=(8, 4))
plt.bar(df['Student'], df['Marks'], color='skyblue', edgecolor='black')
plt.title('Student Marks Distribution')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.ylim(0, 105)
for i, mark in enumerate(df['Marks']):
    plt.text(i, mark + 1, str(mark), ha='center')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 4. Sort marks in descending order and plot again
df_sorted = df.sort_values(by='Marks', ascending=False).reset_index(drop=True)

plt.figure(figsize=(8, 4))
plt.bar(df_sorted['Student'], df_sorted['Marks'], color='coral', edgecolor='black')
plt.title('Student Marks (Sorted in Descending Order)')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.ylim(0, 105)
for i, mark in enumerate(df_sorted['Marks']):
    plt.text(i, mark + 1, str(mark), ha='center')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generate representative data of 500 patient ages (or load: df = pd.read_csv('kaggle_hospital_data.csv'))
np.random.seed(42)
patient_ages = np.random.randint(1, 90, size=500)

df_patients = pd.DataFrame({'Age': patient_ages})

# Visualise age distribution using a histogram
plt.figure(figsize=(9, 5))
counts, bins, patches = plt.hist(df_patients['Age'], bins=range(0, 101, 10),
                                 color='teal', edgecolor='black', alpha=0.75)

plt.title('Patient Age Distribution (Hospital Data)')
plt.xlabel('Age Group')
plt.ylabel('Number of Patients')
plt.xticks(range(0, 101, 10))
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Identify the most common age group
max_idx = np.argmax(counts)
most_common_range = f"{int(bins[max_idx])}-{int(bins[max_idx+1])} years"
print(f"Most common age group: {most_common_range} with {int(counts[max_idx])} patients")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
iris=sns.load_dataset("iris")
print("---Dataset info---")
iris.info()
print("\n---Summary Statistics---")
print(iris.describe())
print("\n---First 5 Rows---")
print(iris.head())
# 3. Scatter plot: Petal Length vs Petal Width colored by Species
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=iris,
    x='petal_length',
    y='petal_width',
    hue='species',
    palette='Set1',
    s=70
)
plt.title("Petal Length vs Petal Width by Species")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()

# 4. Create a Pair Plot
sns.pairplot(iris, hue='species', palette='Set1', corner=False)
plt.suptitle("Pair Plot of Iris Features", y=1.02)
plt.show()
```
